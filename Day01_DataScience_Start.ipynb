```python
print("I am starting Data Science today")
# This is a comment - Python ignores it
name = "Sunday Roy Nweke"
age = 49
goal = "Become a Data Scientist"

print("My name is", name)
print("I am", age, "years old")
print("My goal is to", goal)

# Text (String)
city = "Abuja"
print(type(city))  # Output: <class 'str'>

# Whole number (Integer)
population = 3000000
print(type(population))  # Output: <class 'int'>

# Decimal number (Float)
temperature = 32.5
print(type(temperature))  # Output: <class 'float'>

# True/False (Boolean)
is_student = True
print(type(is_student))  # Output: <class 'bool'>

# Addition
print(10 + 5)         # Output: 15

# Subtraction
print(10 - 5)         # Output: 5

# Multiplication
print(10 * 5)         # Output: 50

# Division (always gives a decimal)
print(10 / 5)         # Output: 2.0

# Power (10 to the power of 2)
print(10 ** 2)        # Output: 100

# Modulo (remainder of division)
print(10 % 3)         # Output: 1
```
